## Objective

This module and notebook is designed to connect two CFDE Data Coordinating Center resources, **MoTrPAC** (exercise multi-omics) and **GTEx** (expression quantitative trait loci, or eQTLs), through a protein language model, **ESM2**, to ask an open scientific question:

> *Is exercise-responsiveness encoded in protein sequence and do the sequence signatures a language model learns correspond to positions where natural genetic variation exists?*

**The three-layers:**
1. **MoTrPAC** → what expression changes (at the gene level) occur across tissues in response to exercise
2. **GTEx** → for those same genes, which regulatory variants (eQTLs) modulate their tissue-specific expression
3. **ESM2** → embed protein sequences, fine-tune to predict exercise-responsiveness, then check whether the sequence positions driving predictions overlap with known eQTL variant positions (see below & the Glossary on what we mean by "embedding").

By the end, you will have built a full, reproducible sequence-to-variant workflow that generalizes to *any* CFDE perturbation dataset.

### A note on data choice: transcriptomics vs. proteomics

Since this module ultimately works with protein sequences (via ESM2), we may ask why we are assessing "exercise-responsive" using transcriptomics (RNA expression) rather than proteomics (protein abundance) data, both of which MoTrPAC provides.

We use transcriptomics here for two practical reasons:
-  MoTrPAC's transcriptomics assay measures substantially more genes than its proteomics assay (roughly 14,000 vs. roughly 9,000 in the liver tissue which is what we're looking at here), giving us a larger, more workable gene set for a teaching example
- It lets us compare directly against a companion proteomics analysis (see the optional addendum) to see whether a gene's RNA-level and protein-level responses actually agree.

The Addendum shows that RNA-level and protein-level responsiveness calls for this tissue and timepoint overlap only about a quarter of the time (roughly 23-26%, depending on which direction you measure the overlap from, we did not compute a correlation between the two layers' effect sizes, just whether each layer's own "responsive" call agrees with the other's).
A gene we call "exercise-responsive" here is responsive *at the RNA level specifically*, but Part 3 and Part 4 model *protein* sequence with ESM2. In other words, the classifier's label comes from the transcriptomic layer, while its input (sequence) comes from the proteomic layer, and those two layers show only weak agreement on which genes are "responsive" (see Addendum for
the full numbers).

### Why ESM2, and not another protein language model

Several protein language models exist (ProtBERT, ProtT5, ProGen, and others). This module uses ESM2 for three practical reasons:

- **Open weights, no gated access**, important for a module meant to run in anyone's Colab session with no setup friction. "Weights" are the millions of learned numbers inside a trained model, the actual result of its training process. "Open weights" means these numbers are freely downloadable by anyone (see Glossary for more).
- **A range of model sizes from one family** (8 million to 15 billion parameters), letting this module use the smallest checkpoint for CPU speed while remaining swappable for a larger one. ESM2 is released as several **checkpoints**, a checkpoint is one specific, saved, trained version of a model, ready to be loaded and used as-is (see Glossary for more).
- **Widely benchmarked**, with strong published performance on structure and variant-effect tasks (Lin et al., 2023, *Science*).

This is a practical choice and does not mean that ESM2 is the only valid option. See:
- ESM2 paper: [Lin et al., 2023, *Science*](https://www.science.org/doi/10.1126/science.ade2574)
- ESM2 code and weights: [github.com/facebookresearch/esm](https://github.com/facebookresearch/esm)
